```python
## api Used: New York Times api Science
import requests
import pandas as pd

# 1. defines the api key and the endpoint url
api_key = "GtWs5GvyNhdmGr5spvvMUUcRUGYZk3LKQl8vPUO35UK2JBxz"
section = "science"
url = f"https://api.nytimes.com/svc/topstories/v2/{section}.json?api-key={api_key}"

# 2. makes the GET request to the NYT api
response = requests.get(url)

# 3. checks if the request was successful
if response.status_code == 200:
    data = response.json()
    # The articles are stored under the 'results' key
    articles_list = data['results']

    # 4. transforms the JSON list directly into a pandas dataframe
    df = pd.DataFrame(articles_list)

    # displays the first few rows and columns
    display(df[['title', 'abstract', 'url', 'published_date']].head())
else:
    print(f"Error: Failed to fetch data. Status code: {response.status_code}")


## quick checks  the shape and basic info of our dataframe
print("Total articles fetched:", len(df))
print(df.info())
```
